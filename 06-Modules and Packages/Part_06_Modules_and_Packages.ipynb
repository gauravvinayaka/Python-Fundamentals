# Part 06: Modules and Packages

## Modules

A module is a Python file that contains code I can reuse.

Python also comes with many built-in modules.

## Importing a Module

```python
import math

print(math.sqrt(25))
print(math.pi)
```

After importing `math`, I can use functions and values stored inside it.

## Importing Specific Items

```python
from math import sqrt, ceil

print(sqrt(49))
print(ceil(4.2))
```

## Using an Alias

```python
import statistics as stats

numbers = [10, 20, 30, 40]

print(stats.mean(numbers))
print(stats.median(numbers))
```

`as stats` gives the module a shorter name.

## A Simple Custom Module Example

```python
import tempfile
import importlib.util
from pathlib import Path

# Create a small Python file temporarily.
with tempfile.TemporaryDirectory() as folder:
    module_file = Path(folder) / "calculator.py"

    module_file.write_text(
        "def add(a, b):\n"
        "    return a + b\n\n"
        "def subtract(a, b):\n"
        "    return a - b\n",
        encoding="utf-8"
    )

    # Load the file as a Python module.
    spec = importlib.util.spec_from_file_location("calculator", module_file)
    calculator = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(calculator)

    print(calculator.add(5, 3))
    print(calculator.subtract(10, 4))
```

A custom module can contain normal functions, classes, or variables that I want to reuse.

## Packages

```python
# A package is usually a folder that contains Python modules.

# Example structure:
#
# my_package/
#     __init__.py
#     calculator.py
#     greetings.py
#
# Then I could import something like:
#
# from my_package.calculator import add
```

Packages help keep a larger project organized by grouping related modules together.

## `__name__` and the Main Guard

```python
print(__name__)

def show_message():
    return "This function can be reused."

if __name__ == "__main__":
    print(show_message())
```

In a normal `.py` file, the code inside the main guard runs when I execute that file directly.

If another file imports it, that block is usually skipped.

## Solution 3

```python
import tempfile
import importlib.util
from pathlib import Path

with tempfile.TemporaryDirectory() as folder:
    module_file = Path(folder) / "converter.py"

    module_file.write_text(
        "def celsius_to_fahrenheit(celsius):\n"
        "    return celsius * 9 / 5 + 32\n\n"
        "def km_to_miles(km):\n"
        "    return km * 0.621371\n",
        encoding="utf-8"
    )

    spec = importlib.util.spec_from_file_location("converter", module_file)
    converter = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(converter)

    print(converter.celsius_to_fahrenheit(25))
    print(round(converter.km_to_miles(10), 2))
```
