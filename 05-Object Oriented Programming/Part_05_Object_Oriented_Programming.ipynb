# Part 05: Object-Oriented Programming

## Object-Oriented Programming

Object-Oriented Programming (OOP) is a way of organizing code using **classes** and **objects**.

I can think of a **class** as a blueprint and an **object** as something created from that blueprint.

## Objects

```python
# Python already has many built-in object types.
number = 10
name = "Gaurav"
colors = ["red", "blue", "green"]

print(type(number))
print(type(name))
print(type(colors))
```

Everything above is an object. I can also create my own type of object by making a class.

## Creating a Class

```python
# A very simple class
class Student:
    pass

student1 = Student()

print(type(student1))
```

`Student` is the class, and `student1` is an instance of that class.

## Attributes

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Aman", 22)

print(student1.name)
print(student1.age)
```

`__init__()` runs automatically when I create an object.

`self.name` and `self.age` are attributes that belong to each object.

## Class Attributes

```python
class Dog:
    species = "mammal"   # Same for every Dog object

    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

dog1 = Dog("Max", "Labrador")
dog2 = Dog("Rocky", "Husky")

print(dog1.name, dog1.breed, dog1.species)
print(dog2.name, dog2.breed, dog2.species)
```

## Methods

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        return f"{self.name} says woof!"

dog1 = Dog("Max")

print(dog1.bark())
```

A method is just a function written inside a class. It usually works with the object's data.

## A Simple Circle Example

```python
class Circle:
    pi = 3.14

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return Circle.pi * self.radius ** 2

    def circumference(self):
        return 2 * Circle.pi * self.radius

circle1 = Circle(3)

print("Area:", circle1.area())
print("Circumference:", circle1.circumference())
```

## Inheritance

```python
class Animal:
    def eat(self):
        return "Eating"

class Dog(Animal):
    def bark(self):
        return "Woof!"

dog1 = Dog()

# Dog can use a method from Animal.
print(dog1.eat())

# Dog also has its own method.
print(dog1.bark())
```

Inheritance lets one class reuse features from another class.

Here, `Dog` inherits from `Animal`.

## Overriding a Method

```python
class Animal:
    def sound(self):
        return "Some animal sound"

class Dog(Animal):
    def sound(self):
        return "Woof!"

animal = Animal()
dog = Dog()

print(animal.sound())
print(dog.sound())
```

## Polymorphism

```python
class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

pets = [Dog(), Cat()]

for pet in pets:
    print(pet.speak())
```

Polymorphism means different objects can use the same method name and give different results.

## Special Methods

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"Book: {self.title}"

    def __len__(self):
        return self.pages

book1 = Book("Python Basics", 200)

print(book1)
print(len(book1))
```

`__str__()` controls what I see when I print the object.

`__len__()` lets me use `len()` on the object.

# Solutions

## Solution 1

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"My name is {self.name} and I am {self.age} years old."

person1 = Person("Riya", 24)
print(person1.introduce())
```

## Solution 2

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return "Withdrawal successful"
        return "Not enough balance"

account = BankAccount("Aman", 500)

account.deposit(200)
print(account.balance)

print(account.withdraw(300))
print(account.balance)
```

## Solution 3

```python
class Vehicle:
    def move(self):
        return "The vehicle is moving."

class Car(Vehicle):
    def move(self):
        return "The car is driving."

vehicle = Vehicle()
car = Car()

print(vehicle.move())
print(car.move())
```
