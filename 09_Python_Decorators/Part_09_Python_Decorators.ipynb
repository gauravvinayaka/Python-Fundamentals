# Part 10: Python Decorators

## Decorators

A decorator lets me add extra behavior to a function without changing the original function itself.

## Functions Can Be Stored in Variables

```python
def say_hello():
    return "Hello"

message_function = say_hello

print(message_function())
```

## A Function Can Be Passed Into Another Function

```python
def say_hello():
    return "Hello"

def run_function(func):
    return func()

print(run_function(say_hello))
```

## A Function Can Return Another Function

```python
def outer():
    def inner():
        return "I am inside outer."
    return inner

new_function = outer()

print(new_function())
```

## A Basic Decorator

```python
def simple_decorator(func):
    def wrapper():
        print("Before the function")
        func()
        print("After the function")
    return wrapper

@simple_decorator
def say_hello():
    print("Hello!")

say_hello()
```

The decorator wraps the original function with some extra code.

## Decorator With Arguments

```python
def show_call(func):
    def wrapper(name):
        print("Function is starting.")
        result = func(name)
        print("Function is finished.")
        return result
    return wrapper

@show_call
def greet(name):
    return f"Hello, {name}"

print(greet("Aman"))
```

## Using `*args` and `**kwargs`

```python
def show_call(func):
    def wrapper(*args, **kwargs):
        print("Running function...")
        return func(*args, **kwargs)
    return wrapper

@show_call
def add(a, b):
    return a + b

print(add(5, 7))
```

`*args` and `**kwargs` make the decorator flexible enough to work with functions that accept different arguments.

## Using `wraps`

```python
from functools import wraps

def simple_decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print("Running:", func.__name__)
        return func(*args, **kwargs)
    return wrapper

@simple_decorator
def multiply(a, b):
    return a * b

print(multiply(3, 4))
print(multiply.__name__)
```

`@wraps` keeps useful information from the original function, such as its name.

# Solutions

## Solution 1

```python
def show_message(func):
    def wrapper():
        print("Function started")
        return func()
    return wrapper

@show_message
def hello():
    print("Hello")

hello()
```

## Solution 2

```python
def show_result(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        print("Result:", result)
        return result
    return wrapper

@show_result
def add(a, b):
    return a + b

add(5, 6)
```

## Solution 3

```python
def repeat_twice(func):
    def wrapper(*args, **kwargs):
        func(*args, **kwargs)
        func(*args, **kwargs)
    return wrapper

@repeat_twice
def say_hi(name):
    print(f"Hi, {name}")

say_hi("Riya")
```
