# Part 11: Python Generators

## Generators

A generator gives values one at a time instead of creating all of them at once.

This can be useful when I am working with a large amount of data.

## Regular Function vs Generator

```python
def regular_numbers():
    return [1, 2, 3]

print(regular_numbers())
```

```python
def generator_numbers():
    yield 1
    yield 2
    yield 3

numbers = generator_numbers()

print(next(numbers))
print(next(numbers))
print(next(numbers))
```

`yield` pauses the function after giving one value.

The next call continues from where it stopped.

## A Generator With a Loop

```python
def count_up_to(limit):
    number = 1

    while number <= limit:
        yield number
        number += 1

for number in count_up_to(5):
    print(number)
```

## A Simple Even Number Generator

```python
def even_numbers(limit):
    for number in range(limit + 1):
        if number % 2 == 0:
            yield number

print(list(even_numbers(10)))
```

## Generator Expressions

```python
squares = (number ** 2 for number in range(1, 6))

print(next(squares))
print(next(squares))
print(list(squares))
```

A generator expression looks similar to a list comprehension, but it uses parentheses instead of square brackets.

## Memory Difference

```python
import sys

list_values = [number for number in range(10000)]
generator_values = (number for number in range(10000))

print("List size:", sys.getsizeof(list_values))
print("Generator size:", sys.getsizeof(generator_values))
```

## `yield from`

```python
def first_group():
    yield 1
    yield 2
    yield 3

def all_numbers():
    yield from first_group()
    yield 4
    yield 5

print(list(all_numbers()))
```

`yield from` lets one generator reuse values from another generator.

# Solutions

## Solution 1

```python
def countdown(start):
    while start >= 0:
        yield start
        start -= 1

print(list(countdown(5)))
```

## Solution 2

```python
def odd_numbers(limit):
    for number in range(1, limit + 1):
        if number % 2 != 0:
            yield number

print(list(odd_numbers(10)))
```

## Solution 3

```python
def square_numbers():
    for number in range(1, 6):
        yield number ** 2

print(list(square_numbers()))
```

## Solution 4

```python
cubes = (number ** 3 for number in range(1, 6))

print(list(cubes))
```
