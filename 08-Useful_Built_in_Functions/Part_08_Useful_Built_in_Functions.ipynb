## Useful Built-in Functions

Python has several useful tools that make it easier to work with lists and other collections.

## `map()`

```python
numbers = [1, 2, 3, 4]

doubled = list(map(lambda x: x * 2, numbers))

print(doubled)
```

`map()` applies the same function to every item.

## `filter()`

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

`filter()` keeps only the items that match a condition.

## `reduce()`

```python
from functools import reduce

numbers = [1, 2, 3, 4]

total = reduce(lambda a, b: a + b, numbers)

print(total)
```

`reduce()` combines the list step by step until one result is left.

## `zip()`

```python
names = ["Aman", "Riya", "Neha"]
scores = [85, 90, 88]

combined = list(zip(names, scores))

print(combined)
```

## `enumerate()`

```python
fruits = ["apple", "banana", "orange"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

`enumerate()` is useful when I need both the index and the item.

## `all()`

```python
values = [10, 20, 30]

result = all(value > 0 for value in values)

print(result)
```

## `any()`

```python
values = [3, 7, 10, 13]

result = any(value % 2 == 0 for value in values)

print(result)
```

## Complex Numbers

```python
number = complex(3, 4)

print(number)
print(number.real)
print(number.imag)
print(abs(number))
```
