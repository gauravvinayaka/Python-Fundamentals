# Part 07: Errors and Exception Handling

## Errors and Exceptions

Errors happen when Python cannot complete an operation.

`try` and `except` let me handle some errors without stopping the entire program.

## Basic `try` and `except`

```python
try:
    answer = 10 / 0
except ZeroDivisionError:
    print("I cannot divide by zero.")
```

## Handling Invalid Input

```python
value = "hello"

try:
    number = int(value)
    print(number)
except ValueError:
    print("That value cannot be converted to an integer.")
```

## `else`

```python
text = "25"

try:
    number = int(text)
except ValueError:
    print("Invalid number")
else:
    print("The number is:", number)
```

`else` runs only if the `try` block finishes without an exception.

## `finally`

```python
try:
    file_message = "Pretend I opened a file here."
    print(file_message)
finally:
    print("This part always runs.")
```

`finally` runs whether there is an error or not.

## Handling More Than One Error

```python
values = [10, 20, 30]

try:
    index = int("5")
    print(values[index])
except ValueError:
    print("The index was not a valid integer.")
except IndexError:
    print("That index does not exist in the list.")
```

## Raising an Error

```python
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")
    return f"Age is {age}"

print(check_age(25))
```

`raise` lets me create an error intentionally when a value is not acceptable.

## A Simple Custom Exception

```python
class LowBalanceError(Exception):
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise LowBalanceError("Not enough money in the account.")
    return balance - amount

print(withdraw(500, 200))
```

## Basic Unit Testing

```python
import unittest

def add(a, b):
    return a + b

class TestAdd(unittest.TestCase):
    def test_add_positive_numbers(self):
        self.assertEqual(add(2, 3), 5)

    def test_add_negative_numbers(self):
        self.assertEqual(add(-2, -3), -5)

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestAdd)
result = unittest.TextTestRunner(verbosity=1).run(suite)

print("Passed:", result.wasSuccessful())
```

# Solutions

## Solution 1

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

## Solution 2

```python
def check_positive(number):
    if number < 0:
        raise ValueError("Number cannot be negative.")
    return number

print(check_positive(5))
```

## Solution 3

```python
class InvalidScoreError(Exception):
    pass

def check_score(score):
    if score < 0 or score > 100:
        raise InvalidScoreError("Score must be between 0 and 100.")
    return score

print(check_score(85))
```

```python
import unittest

def multiply(a, b):
    return a * b

class TestMultiply(unittest.TestCase):
    def test_positive_numbers(self):
        self.assertEqual(multiply(3, 4), 12)

    def test_with_zero(self):
        self.assertEqual(multiply(8, 0), 0)

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestMultiply)
unittest.TextTestRunner(verbosity=1).run(suite)
```
